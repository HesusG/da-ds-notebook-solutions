```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LinearRegression
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

Cargamos los datos

df = pd.read_csv('/datasets/users_behavior.csv')

Arbol de Decisión de Clasificación

```python
df_2 = df.copy()
features_1 = df_2.drop(['is_ultra'],axis=1)
target_1 = df_2['is_ultra']
```

```python
features_train_1, features_valid_1, target_train_1, target_valid_1 = train_test_split(features_1, target_1, test_size=.25,random_state=12345)
```

```python
best_model = None
best_result = 0
best_depth = 0
for depth in range(1, 6): # selecciona el rango del hiperparámetro, el límite superior no se incluye
    model = DecisionTreeClassifier(random_state=12345, max_depth=depth)
    model.fit(features_train_1,target_train_1) # entrena el modelo en el conjunto de entrenamiento

    
    predictions_valid_1 = model.predict(features_valid_1) # obtén las predicciones del modelo en el conjunto de validación
    result = accuracy_score(target_valid_1, predictions_valid_1)
    
    if result > best_result:
        best_model = model
        best_result = result
        best_depth = depth

print(f"Accuracy del mejor árbol de decisión (max_depth = {best_depth}): {best_result:.4f}")
```

En nuestro primer modelo el hiperparametro que se utilizo fue el de profundidad ajustando en 5, puesto que a mayor profunidad el modelo tiende a sobreajustarse.

Bosque Aleatoria de Regresión

```python
df_3 = df.copy()
df_train_1, df_valid_1 = train_test_split(df_3, random_state=54321, test_size =.25)
```

```python
features_train_2 = df_train_1.drop(['is_ultra'],axis=1)
target_train_2 = df_train_1['is_ultra']
features_valid_2 = df_valid_1.drop(['is_ultra'],axis=1)
target_valid_2 = df_valid_1['is_ultra']
```

```python
best_model_2 = None
best_result_2 = 0
best_est_ = 0
best_depth_2 = 0
for est in range(10, 51, 10):
    for depth in range (1, 11):
        model = RandomForestClassifier(random_state=54321, n_estimators=est, max_depth=depth)# inicializa el constructor de modelos con los parámetros random_state=54321, n_estimators=est y max_depth=depth
        model.fit(features_train_2,target_train_2) # entrena el modelo en el conjunto de entrenamiento
        predictions_valid = model.predict(features_valid_2) # obtén las predicciones del modelo en el conjunto de validación
        
        result = accuracy_score(target_valid_2,predictions_valid)# calcula la RECM en el conjunto de validación
        print(f"Accuracy para n_estimators =", est, "depth =", depth,":",result)
        if result > best_result_2: 
            best_result_2 = result
            best_est_2 = est
            best_depth_2 = depth
            best_model_2 = model
            

print("El mejor Random Forest: accuracy =", best_result_2, "n_estimators =",best_est_2, "max_depth:", best_depth_2)
```

Para el segundo modelo utilizamos dos hiperparametros, n_estimators para ver el numero de arboles a utilizar y depth en 9 para ajustar la profunidad maxima de cada arbol

REGRESIÓN LOGISITICA

```python
df_4 = df.copy()
```

```python
df_train_4, df_valid_4 = train_test_split(df_4, test_size=.25, random_state=54321) # segmenta el 25% de los datos para hacer el conjunto de validación

features_train_4 = df_train_4.drop(['is_ultra'], axis=1)
target_train_4 = df_train_4['is_ultra']
features_valid_4 = df_valid_4.drop(['is_ultra'], axis=1)
target_valid_4 = df_valid_4['is_ultra']

model = LogisticRegression(random_state=54321, solver='liblinear')# inicializa el constructor de regresión logística con los parámetros random_state=54321 y solver='liblinear'
model.fit(features_train_4,target_train_4) # entrena el modelo en el conjunto de entrenamiento
score_train = model.score(features_train_4,target_train_4) # calcula la puntuación de accuracy en el conjunto de entrenamiento
score_valid = model.score(features_valid_4,target_valid_4) # calcula la puntuación de accuracy en el conjunto de validación

print("Accuracy del modelo de regresión logística en el conjunto de entrenamiento:", score_train)
print("Accuracy del modelo de regresión logística en el conjunto de validación:", score_valid)
```

Para el tercer modelo utilizamos un random state y el solver de 'liblinear' pues no hay demasiadas columnas.

Para cada modelo utilizado a la hora de partir los datos de entrenamiento y de validación utilizamos la partición 3:1:1
