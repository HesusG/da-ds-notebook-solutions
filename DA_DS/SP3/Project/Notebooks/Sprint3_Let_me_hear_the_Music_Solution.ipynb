### Hypothesis: comparing user behavior in two cities <a id='activity'></a>

The hypothesis states that there are differences in the way that music is consumed by users in Springfield and Shelbyville. To test this, use the data from three days of the week: Monday, Wednesday, and Friday.

* Group the users by city.
* Compare the number of tracks each group played on Monday, Wednesday, and Friday.


Perform each computation separately.

The first step is to evaluate user activity in each city. Remember about the split-apply-combine stages we talked about previously in the lesson. Your goal now is to group the data by city, apply the appropriate method for counting during the apply stage, and then find the number of songs played in each group by specifying the column to get the count for.

Here is an example of how the final result should look like:
`df.groupby(by='....')['column'].method()`

```python
# Counting up the tracks played in each city
df.groupby(by='city')['track'].count()
```

As a result, we see that Springfield has more tracks played than Shelbyville. However, this does not necessarily mean that citizens of Springfield listen to music more frequently. The possible larger size of a city simply results in a greater number of users.

Now let's group the data by day of the week and find the number of tracks played on Monday, Wednesday, and Friday. Use the same approach as before but now we need different grouping.


```python
# Calculating tracks played on each day of the week
df.groupby(by=['day'])['track'].count()
```

As a result we see that Wednesday is the quietest day overall. However, if we consider the two cities separately, we might reach a different conclusion.

You have learned how to count entries by grouping them by city or day. Now, you need to write a function that can count entries based on both criteria simultaneously.

Create the `number_tracks()` function to calculate the number of songs played for a given day **and** city. The function should accept two parameters:

- `day`: a day of the week to filter by. For instance, `'Monday'`.
- `city`: a city to filter by. For instance, `'Springfield'`.

Within the function, you will apply consecutive filtering with logical indexing.

First, filter the data by day and then filter the resulting table by city.

After filtering the data by two criteria, count the number of 'user_id' column values in the resulting table. This count represents the number of entries you are looking for. Store the result in a new variable and return it from the function.

```python
# <creating the function number_tracks()>
# Declare the function number_tracks() with two parameters: day= and city=.
def number_tracks(day, city):
    # Store the DataFrame rows where the value in the 'day' column is equal to the day= parameter
    track_list = df[df['day'] == day]

    # Filter the rows where the value in the 'city' column is equal to the city= parameter
    track_list = track_list[track_list['city'] == city]

    # Extract the 'user_id' column from the filtered table and apply the count() method
    track_list_count = track_list['user_id'].count()

    # Return the number of 'user_id' column values
    return track_list_count
```

Call `number_tracks()` six times, changing the parameter values, so that you retrieve the data on both cities for each of the three days.

```python
# the number of songs played in Springfield on Monday
print(number_tracks('Monday', 'Springfield'))
```

```python
# the number of songs played in Shelbyville on Monday
print(number_tracks('Monday', 'Shelbyville'))
```

```python
# the number of songs played in Springfield on Wednesday
print(number_tracks('Wednesday', 'Springfield'))
```

```python
# the number of songs played in Shelbyville on Wednesday
print(number_tracks('Wednesday', 'Shelbyville'))
```

```python
# the number of songs played in Springfield on Friday
print(number_tracks('Friday', 'Springfield'))
```

```python
# the number of songs played in Shelbyville on Friday
print(number_tracks('Friday', 'Shelbyville'))
```

Make your conclusions about the hypothesis that you test.

**Conclusions**

The data reveals differences in user behavior:

- In Springfield, the number of songs played peaks on Mondays and Fridays, while on Wednesday there is a decrease in activity.
- In Shelbyville, on the contrary, users listen to music more on Wednesday. User activity on Monday and Friday is smaller.

So the hypothesis seems to be correct. We accept it.
